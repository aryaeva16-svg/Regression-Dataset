```python
import random
import numpy as np

num_tasks = int(input("Enter number of tasks: "))
num_processors = int(input("Enter number of processors: "))
num_buses = int(input("Enter number of buses: "))
```

```python
task_table = np.array([
    [200, 1, 10],   # T1
    [300, 2, 15],   # T2
    [150, 1, 12],   # T3
    [400, 3, 20]    # T4
])

print("\n--- TASK TABLE ---")
print(task_table)
```

```python
buses = {
    "B1": 50,
    "B2": 100
}

print("\n--- BUS DATABASE ---")
for b in buses:
    print(f"{b}: {buses[b]} MB/s")
```

```python
processors = {
    1: {"type": "GPP", "MIPS": 100, "buses": ["B1"]},
    2: {"type": "HPP", "MIPS": 200, "buses": ["B1", "B2"]},
    3: {"type": "DSP", "MIPS": 150, "buses": ["B2"]}
}

print("\n--- PROCESSOR LIBRARY ---")
for p in processors:
    print(f"P{p}: {processors[p]}")
```

```python
data_transfer = {
    (0,1): 40,
    (0,2): 60,
    (1,3): 50,
    (2,3): 30
}
connectivity_matrix = np.zeros((num_tasks, num_tasks), dtype=int)
data_matrix = np.zeros((num_tasks, num_tasks))

for (src, dst), val in data_transfer.items():
    connectivity_matrix[src][dst] = 1
    data_matrix[src][dst] = val

print("\n--- CONNECTIVITY MATRIX ---")
print(connectivity_matrix)

print("\n--- DATA TRANSFER MATRIX ---")
print(data_matrix)
```

```python
feasibility_matrix = np.zeros((num_tasks, num_processors), dtype=int)

for t in range(num_tasks):
    MI = task_table[t][0]

    for p in range(1, num_processors+1):
        p_type = processors[p]["type"]

        if p_type in ["GPP", "HPP"]:
            feasibility_matrix[t][p-1] = 1
        elif p_type == "DSP":
            if MI <= 200:
                feasibility_matrix[t][p-1] = 1

print("\n--- FEASIBILITY MATRIX ---")
print(feasibility_matrix)
```

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.DiGraph()

# Add nodes
for i in range(num_tasks):
    G.add_node(f"T{i+1}")

# Add edges with data labels
for (src, dst), data in data_transfer.items():
    G.add_edge(f"T{src+1}", f"T{dst+1}", weight=data)

pos = nx.spring_layout(G)

edge_labels = nx.get_edge_attributes(G, 'weight')

plt.figure()
nx.draw(G, pos, with_labels=True, node_size=2000)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Task Graph with Data Transfer")
plt.show()
```

```python
def compute_fitness(chromosome):
    total_time = 0
    penalty = 0
    finish_time = np.zeros(num_tasks)

    for t in range(num_tasks):
        if feasibility_matrix[t][chromosome[t]-1] == 0:
            return 10000, finish_time

    for t in range(num_tasks):
        MI = task_table[t][0]
        MIPS = processors[chromosome[t]]["MIPS"]
        exec_time = MI / MIPS
        total_time += exec_time
        finish_time[t] += exec_time

    for i in range(num_tasks):
        for j in range(num_tasks):
            if connectivity_matrix[i][j] == 1:
                if chromosome[i] != chromosome[j]:

                    p1 = processors[chromosome[i]]
                    p2 = processors[chromosome[j]]

                    common = list(set(p1["buses"]) & set(p2["buses"]))

                    if common:
                        rate = max([buses[b] for b in common])
                        comm_time = data_matrix[i][j] / rate
                        total_time += comm_time
                        finish_time[j] += comm_time
                    else:
                        penalty += 1000

    for t in range(num_tasks):
        min_t = task_table[t][1]
        max_t = task_table[t][2]

        if finish_time[t] < min_t:
            penalty += 50
        if finish_time[t] > max_t:
            penalty += 200

    return total_time + penalty, finish_time
```

```python
def create_chromosome():
    chromosome = []
    for t in range(num_tasks):
        possible = [p+1 for p in range(num_processors) if feasibility_matrix[t][p] == 1]
        chromosome.append(random.choice(possible))
    return chromosome

def selection(pop):
    pop = sorted(pop, key=lambda x: compute_fitness(x)[0])
    return pop[:2]

def crossover(p1, p2):
    point = random.randint(1, num_tasks-1)
    return p1[:point] + p2[point:]

def mutate(chromosome, rate=0.2):
    if random.random() < rate:
        idx = random.randint(0, num_tasks-1)
        old = chromosome[idx]

        possible = [p+1 for p in range(num_processors) if feasibility_matrix[idx][p] == 1]
        chromosome[idx] = random.choice(possible)

        print(f"Mutation: T{idx+1} P{old}→P{chromosome[idx]}")

    return chromosome
```

```python
def GA(generations=20, pop_size=6):
    pop = [create_chromosome() for _ in range(pop_size)]

    for gen in range(generations):
        new_pop = []

        for _ in range(pop_size):
            p1, p2 = selection(pop)
            child = crossover(p1, p2)
            child = mutate(child)
            new_pop.append(child)

        pop = new_pop

        best = min(pop, key=lambda x: compute_fitness(x)[0])
        print(f"Gen {gen+1}: Best Fitness = {compute_fitness(best)[0]}")

    best = min(pop, key=lambda x: compute_fitness(x)[0])
    return best, compute_fitness(best)
```

```python
def GA(generations=20, pop_size=6):
    pop = [create_chromosome() for _ in range(pop_size)]

    for gen in range(generations):
        new_pop = []

        for _ in range(pop_size):
            p1, p2 = selection(pop)
            child = crossover(p1, p2)
            child = mutate(child)
            new_pop.append(child)

        pop = new_pop

        best = min(pop, key=lambda x: compute_fitness(x)[0])
        print(f"Gen {gen+1}: Best Fitness = {compute_fitness(best)[0]}")

    best = min(pop, key=lambda x: compute_fitness(x)[0])
    return best, compute_fitness(best)
```

```python
best_sol, (best_time, finish_time, comp_time, comm_time) = GA()

print("\n--- FINAL RESULT ---")
print("Best Chromosome:", best_sol)
print("Total Time:", best_time)

print("\n--- TIME BREAKDOWN ---")
print(f"Computation Time: {comp_time}")
print(f"Communication Time: {comm_time}")

print("\n--- TASK MAPPING ---")
for i in range(num_tasks):
    print(f"T{i+1} -> P{best_sol[i]}")

print("\n--- DEADLINE CHECK ---")
for t in range(num_tasks):
    ft = finish_time[t]
    min_t = task_table[t][1]
    max_t = task_table[t][2]

    if ft < min_t:
        status = "Too Early"
    elif ft > max_t:
        status = "Deadline Missed"
    else:
        status = "OK"

    print(f"T{t+1}: Finish={round(ft,2)}, Range=({min_t}-{max_t}) → {status}")
```
